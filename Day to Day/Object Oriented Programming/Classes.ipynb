```python
class Rectangle:
    def __init__(self, width, height):
        self.set_width(width)
        self.set_height(height)

    def get_width(self):
        return self._width

    def set_width(self, width):
        if width <= 0:
            raise ValueError("Width must be greater than 0")
        self._width = width

    def get_height(self):
        return self._height

    def set_height(self, height):
        if height <= 0:
            raise ValueError("Height must be greater than 0")
        self._height = height

    def area(self):
        return self._width * self._height

    def perimeter(self):
        return 2 * (self._width + self._height)

    def to_string(self):
        return f"Rectangle: width = {self._width}, height = {self._height}"

    def __str__(self):
        return self.to_string()

    def __repr__(self):
        return self.to_string()

    def __eq__(self, other):
        return (
            isinstance(other, Rectangle)
            and self._width == other._width
            and self._height == other._height
        )


r1 = Rectangle(10, 20)
r2 = Rectangle(10, 20)

r1.get_width()

r1.width #this will get following error as there is no public attribute width in the Rectangle class. The width attribute is private and can only be accessed through the get_width() method. To get the width of the rectangle, you can use the following code:
"""ArithmeticError---------------------------------------------------------------------------
AttributeError                            Traceback (most recent call last)
Cell In[29], line 1
----> 1 class Rectangle:
      2     def __init__(self, width, height):
      3         self._width = width

Cell In[29], line 41, in Rectangle()
     38 r1 = Rectangle(10,20)
     39 r2 = Rectangle(10,20)
---> 41 r1.width

AttributeError: 'Rectangle' object has no attribute 'width' """
# this will print the width of the rectangle, which is 10
```

```python
# This is a more complete implementation of the Student class, it includes validation for the name and house, it also includes error handling for the get_students function, it raises a ValueError if the name or house is missing or if the house is not valid. This way we can ensure that the data we are storing in our class is valid and we can also handle errors gracefully.
class Student:
    def __init__(self, name: str, age: int, email: str):
        name = name.strip()
        email = email.strip()

        if not name:
            raise ValueError("Name cannot be empty")

        if age < 5 or age > 120:
            raise ValueError("Age must be between 5 and 120")

        if "@" not in email or "." not in email:
            raise ValueError("Email looks invalid")

        self.name = name
        self.age = age
        self.email = email


def main():
    name = input("Enter name: ")
    age = int(input("Enter age: "))
    email = input("Enter email: ")
    
    try:
        s = Student(name,age,email )
        print(f"Student is registered with Name: {s.name} Age: ({s.age})and Email -> {s.email}")
    except ValueError as e:
        # here whatever the error message was in the ValueError, 
        # it will be captured in variable e and we can print it out or log it
        print("Could not create student:", e)


if __name__ == "__main__":
    main()
```
